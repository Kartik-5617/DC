```python

A = {'x1': 0.2, 'x2': 0.7, 'x3': 0.5}
B = {'x1': 0.6, 'x2': 0.4, 'x3': 0.8}

print("Fuzzy Set A:", A)
print("Fuzzy Set B:", B)
```

```python
# Fuzzy Union
def fuzzy_union(A, B):
    return {x: max(A.get(x, 0), B.get(x, 0)) for x in set(A) | set(B)}

# Fuzzy Intersection
def fuzzy_intersection(A, B):
    return {x: min(A.get(x, 0), B.get(x, 0)) for x in set(A) & set(B)}

# Fuzzy Complement
def fuzzy_complement(A):
    return {x: 1 - A[x] for x in A}

# Fuzzy Difference
def fuzzy_difference(A, B):
    return {x: min(A.get(x, 0), 1 - B.get(x, 0)) for x in A}
```

```python
union_result = fuzzy_union(A, B)
intersection_result = fuzzy_intersection(A, B)
complement_A = fuzzy_complement(A)
difference_result = fuzzy_difference(A, B)

print("Union A ∪ B:", union_result)
print("Intersection A ∩ B:", intersection_result)
print("Complement of A:", complement_A)
print("Difference A - B:", difference_result)
```

```python
def cartesian_product(A, B):
    return {(x, y): min(A[x], B[y]) for x in A for y in B}

R1 = cartesian_product(A, B)
R2 = cartesian_product(B, A)

print("Fuzzy Relation R1 (A × B):")
for k, v in R1.items():
    print(f"{k}: {v}")

print("\nFuzzy Relation R2 (B × A):")
for k, v in R2.items():
    print(f"{k}: {v}")
```

```python
def max_min_composition(R1, R2):
    result = {}
    X = set(x for (x, _) in R1)
    Z = set(z for (_, z) in R2)
    Y_common = set(y for (_, y) in R1) & set(y for (y, _) in R2)

    for x in X:
        for z in Z:
            mins = [min(R1.get((x, y), 0), R2.get((y, z), 0)) for y in Y_common]
            result[(x, z)] = max(mins) if mins else 0
    return result

composition_result = max_min_composition(R1, R2)

print("Max-Min Composition (R1 ○ R2):")
for k, v in composition_result.items():
    print(f"{k}: {v}")
```
